## Employees data dataset:

- This dataset contains 50 samples designed for learning and experimentation with multiple linear regression models.

## Each record represents an individual with:

- Age
- Years of work experience
- Annual income (generated using a linear relationship with realistic noise)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("employee_income.csv")
```

```python
df
```

## Goal 1: Relation between the work experience and income

```python
Income = df.sort_values(by = "income", ascending = False)
```

```python
Income[['experience', 'income']]
```

```python
plt.figure(figsize=(10, 30))
sns.scatterplot(data = Income, x = Income['experience'], y = Income['income'], palette = 'Blues',
               hue = Income['income'])
plt.xlabel("Experience", size = 15)
plt.ylabel("Income", size = 15)
plt.title("Experience vs income", size = 15)
plt.show()
```

## Insights:
- There is a positive relationship between the work experience and employee income.
- Employees with 14, 12, 11, 13, 10, 7 years of experience earn the highest income.
- Employees with 0 to 1 year of experience, 2 years of experience, 3 years of experience earn the least income.
## Conclusions:
- The more the work experience, the higher the income.

## Goal 2: Relation between age and experience

```python
Age_experience = df[['age', 'experience']]
```

```python
Age_experience
```

```python
Age_experience_sort = Age_experience.sort_values(by = "experience", ascending = False)
```

```python
Age_experience_sort
```

```python
plt.figure(figsize = (10, 10))
sns.scatterplot(data = Age_experience_sort, x = Age_experience_sort['experience'], 
                y = Age_experience_sort['age'], palette = 'Blues',
               hue = Income['experience'])
plt.xlabel("Experience", size = 15)
plt.ylabel("Age", size = 15)
plt.title("Experience vs Age", size = 15)
plt.show()
```

## Insights:
- There is a variable relationship between the work experience and age.
- Employees with the highest experience lies in the age range between 23 and 50 years old.
- While the oldest employees in the age range between 56 and 59 years old have 7 years of experience.
## Conclusions:
- While age is an important factor in work experience, but work experience comes from knowledge and applications: two important factors that shape work experience.
